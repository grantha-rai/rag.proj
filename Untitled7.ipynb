```python
# ==========================================
# MOVIE RECOMMENDATION SYSTEM USING RAG
# ==========================================

# Install required library
!pip -q install scikit-learn

# Import libraries
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# ------------------------------------------
# 1. MOVIE KNOWLEDGE BASE
# ------------------------------------------

movies = [
    {
        "title": "Interstellar",
        "genre": "Science Fiction, Adventure, Drama",
        "info": "A space adventure about astronauts, time, family and exploring another galaxy."
    },
    {
        "title": "Inception",
        "genre": "Science Fiction, Action, Thriller",
        "info": "A mind-bending story about dreams, the human mind and a team trying to perform a difficult mission."
    },
    {
        "title": "The Dark Knight",
        "genre": "Action, Crime, Thriller",
        "info": "Batman faces the Joker in a dark story involving crime, action and justice."
    },
    {
        "title": "3 Idiots",
        "genre": "Comedy, Drama, Friendship",
        "info": "Three engineering students experience friendship, college life, education and following their dreams."
    },
    {
        "title": "Dangal",
        "genre": "Sports, Drama",
        "info": "A father trains his daughters in wrestling and supports them through their journey to success."
    },
    {
        "title": "Taare Zameen Par",
        "genre": "Drama, Family",
        "info": "A child struggles with learning difficulties and a teacher helps him discover his talents."
    },
    {
        "title": "KGF",
        "genre": "Action, Drama",
        "info": "A powerful action story about ambition, power, poverty and a man's rise to become a powerful figure."
    },
    {
        "title": "Avengers Endgame",
        "genre": "Action, Science Fiction, Superhero",
        "info": "The Avengers come together to fight against Thanos and try to save the universe."
    },
    {
        "title": "The Matrix",
        "genre": "Science Fiction, Action",
        "info": "A computer programmer discovers that the world around him is actually a simulated reality."
    },
    {
        "title": "Forrest Gump",
        "genre": "Drama, Romance",
        "info": "A kind-hearted man experiences friendship, love and important events throughout his life."
    }
]

# ------------------------------------------
# 2. CREATE DOCUMENTS
# ------------------------------------------

documents = []

for movie in movies:
    document = (
        movie["title"] + " " +
        movie["genre"] + " " +
        movie["info"]
    )
    documents.append(document)

# ------------------------------------------
# 3. CREATE VECTOR DATABASE
# ------------------------------------------

vectorizer = TfidfVectorizer()

movie_vectors = vectorizer.fit_transform(documents)

# ------------------------------------------
# 4. USER QUERY
# ------------------------------------------

print("===================================")
print("     🎬 MOVIE RECOMMENDATION AI")
print("===================================")

question = input("\nWhat type of movie are you looking for? ")

# ------------------------------------------
# 5. RETRIEVAL
# ------------------------------------------

question_vector = vectorizer.transform([question])

similarity = cosine_similarity(
    question_vector,
    movie_vectors
)[0]

# Get top 3 movies
top_movies = similarity.argsort()[-3:][::-1]

# ------------------------------------------
# 6. GENERATE ANSWER
# ------------------------------------------

print("\n🍿 RECOMMENDED MOVIES")
print("------------------------------")

for position, index in enumerate(top_movies, 1):

    movie = movies[index]

    print(f"\n{position}. 🎬 {movie['title']}")
    print(f"   Genre: {movie['genre']}")
    print(f"   Why: {movie['info']}")

print("\n--------------------------------")
print("RAG process completed successfully!")
```
